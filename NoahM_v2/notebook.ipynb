```python
dossier_racine = Path(f"./data/processing/")
AUTOTUNE = tf.data.experimental.AUTOTUNE

def create_datasets(img_size):

    train_set = tf.keras.preprocessing.image_dataset_from_directory(
        dossier_racine,
        validation_split=1-train_split,
        subset="training",
        seed=42,
        batch_size=BATCH_SIZE,
        image_size=(img_size, img_size)
    )

    test_set = tf.keras.preprocessing.image_dataset_from_directory(
        dossier_racine,
        validation_split=test_split,
        subset="validation",
        seed=42,
        batch_size=BATCH_SIZE,
        image_size=(img_size, img_size)
    )

    train_set = train_set.cache().shuffle(1000).prefetch(
        buffer_size=AUTOTUNE
    )
    test_set = test_set.cache().prefetch(
        buffer_size=AUTOTUNE
    )

    return train_set, test_set
```

```python
# Etape 2 : Définir différent modele selon les résolutions et amélioration
```

```python
def create_cnn(size, dense):
    model_layer = [
        layers.Rescaling(1./255, input_shape=(size, size, 3)),

        layers.Conv2D(32, 3, activation="relu"),
        layers.MaxPooling2D(),

        layers.Conv2D(64, 3, activation="relu"),
        layers.MaxPooling2D(),
    ]

    if size >= 256:
        model_layer.extend([
            layers.Conv2D(128, 3, activation="relu"),
            layers.MaxPooling2D(),
        ])

    model_layer.extend([
        layers.Flatten(),
        layers.Dense(dense, activation="relu"),
        layers.Dense(1, activation="sigmoid")
    ])

    model = Sequential(model_layer)

    model.compile(
        optimizer="adam",
        loss="binary_crossentropy",
        metrics=["accuracy"]
    )

    return model


models = {}

for dense in dense_layers:
    for size in resolutions:
        models[(size, dense)] = create_cnn(size, dense)
models
```

```python
# Etape 3 : Entrainement des modeles
```

```python
Epoch = 10
histories = {}

for dense in dense_layers:
    for size in resolutions:

        print(f"Training : résolution={size}, dense={dense}")

        train_set, test_set = create_datasets(size)

        history = models[(size, dense)].fit(
            train_set,
            validation_data=test_set,
            epochs=Epoch
        )

        histories[(size, dense)] = history
```

```python
for (size, dense), history in histories.items():

    acc = history.history["accuracy"]
    val_acc = history.history["val_accuracy"]

    loss = history.history["loss"]
    val_loss = history.history["val_loss"]

    epochs_range = range(1, len(acc) + 1)

    plt.figure(figsize=(12, 5))

    # Accuracy
    plt.subplot(1, 2, 1)
    plt.plot(epochs_range, acc, label="Training Accuracy")
    plt.plot(epochs_range, val_acc, label="Validation Accuracy")
    plt.legend()
    plt.title(f"Accuracy - {size}x{size}, Dense={dense}")
    plt.xlabel("Epoch")
    plt.ylabel("Accuracy")

    # Loss
    plt.subplot(1, 2, 2)
    plt.plot(epochs_range, loss, label="Training Loss")
    plt.plot(epochs_range, val_loss, label="Validation Loss")
    plt.legend()
    plt.title(f"Loss - {size}x{size}, Dense={dense}")
    plt.xlabel("Epoch")
    plt.ylabel("Loss")

    plt.show()
```

```python
best_model_key = None
best_val_accuracy = 0

for key, history in histories.items():

    max_val_accuracy = max(history.history["val_accuracy"])

    if max_val_accuracy > best_val_accuracy:
        best_val_accuracy = max_val_accuracy
        best_model_key = key

print("Meilleur modèle :")
print(f"Résolution : {best_model_key[0]}x{best_model_key[0]}")
print(f"Dense : {best_model_key[1]}")
print(f"Validation accuracy : {best_val_accuracy:.4f}")
```
