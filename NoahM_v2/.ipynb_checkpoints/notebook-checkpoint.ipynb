```python
dossier_racine = Path(f"./data/processing/")

def create_datasets(img_size):

    train_set = tf.keras.preprocessing.image_dataset_from_directory(
        dossier_racine,
        validation_split=1-train_split,
        subset="training",
        seed=42,
        batch_size=BATCH_SIZE,
        image_size=(img_size, img_size)
    )

    test_set = tf.keras.preprocessing.image_dataset_from_directory(
        dossier_racine,
        validation_split=test_split,
        subset="validation",
        seed=42,
        batch_size=BATCH_SIZE,
        image_size=(img_size, img_size)
    )

    train_set = train_set.cache().shuffle(1000).prefetch(
        buffer_size=AUTOTUNE
    )
    test_set = test_set.cache().prefetch(
        buffer_size=AUTOTUNE
    )

    return train_set, test_set
```

```python
# Etape 2 : Définir différent modele selon les résolutions et amélioration
```

```python
def create_cnn(size, dense):
    model_layer = [
        layers.Rescaling(1./255, input_shape=(size, size, 3)),

        layers.Conv2D(32, 3, activation="relu"),
        layers.MaxPooling2D(),

        layers.Conv2D(64, 3, activation="relu"),
        layers.MaxPooling2D(),
    ]

    if size >= 256:
        model_layer.extend([
            layers.Conv2D(128, 3, activation="relu"),
            layers.MaxPooling2D(),
        ])

    model_layer.extend([
        layers.Flatten(),
        layers.Dense(dense, activation="relu"),
        layers.Dense(1, activation="sigmoid")
    ])

    model = Sequential(model_layer)

    model.compile(
        optimizer="adam",
        loss="binary_crossentropy",
        metrics=["accuracy"]
    )

    return model


models = {}

for dense in dense_layers:
    for size in resolutions:
        models[(size, dense)] = create_cnn(size, dense)
models
```

```python
# Etape 3 : Entrainement des modeles
```

```python
Epoch = 10
histories = {}

for dense in dense_layers:
    for size in resolutions:

        print(f"Training : résolution={size}, dense={dense}")

        train_set, test_set = create_datasets(size)

        history = models[(size, dense)].fit(
            train_set,
            validation_data=test_set,
            epochs=Epoch
        )

        histories[(size, dense)] = history
```
